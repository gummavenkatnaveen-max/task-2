```python
df = df.drop_duplicates()
```

```python
df["Created_Date"] = pd.to_datetime(
    df["Created_Date"],
    errors="coerce"
)

df["Resolved_Date"] = pd.to_datetime(
    df["Resolved_Date"],
    errors="coerce"
)
```

```python
df[["Created_Date", "Resolved_Date"]].dtypes
```

```python
df["Region"] = df["Region"].fillna("Unknown")
```

```python
df["Channel"] = df["Channel"].fillna("Unknown")
```

```python
df["Customer_Satisfaction"].isnull().sum()
```

```python
df["Customer_Satisfaction"].mean()
```

```python
df.loc[
    df["Status"] == "Pending",
    "Resolved_Date"
] = pd.NaT
```

```python
missing_resolved = df[
    df["Status"].isin(["Resolved", "Closed"]) &
    df["Resolved_Date"].isnull()
]

missing_resolved[
    [
        "Ticket_ID",
        "Status",
        "Created_Date",
        "Resolution_Hours"
    ]
]
```

```python
mask = (
    df["Status"].isin(["Resolved", "Closed"]) &
    df["Resolved_Date"].isnull() &
    df["Resolution_Hours"].notnull()
)

df.loc[mask, "Resolved_Date"] = (
    df.loc[mask, "Created_Date"] +
    pd.to_timedelta(
        df.loc[mask, "Resolution_Hours"],
        unit="h"
    )
)
```

```python
df["Resolution_Hours_Calculated"] = (
    df["Resolved_Date"] -
    df["Created_Date"]
).dt.total_seconds() / 3600
```

```python
df.loc[
    df["Status"] == "Pending",
    "Resolution_Hours_Calculated"
] = np.nan
```

```python
df["Resolution_Hours"] = df["Resolution_Hours_Calculated"]
```

```python
df.drop(
    columns=["Resolution_Hours_Calculated"],
    inplace=True
)
```

```python
df[df["Resolution_Hours"] < 0]
```

```python
df["Resolution_Hours"].describe()
```

```python
df["SLA_Status"] = np.where(
    df["Status"] == "Pending",
    "Pending",
    np.where(
        df["Resolution_Hours"] <= df["SLA_Hours"],
        "Met",
        "Breached"
    )
)
```

```python
total_tickets = df["Ticket_ID"].nunique()

print("Total Tickets:", total_tickets)
```

```python
resolved_tickets = df[
    df["Status"].isin(["Resolved", "Closed"])
]["Ticket_ID"].nunique()

print("Resolved/Closed Tickets:", resolved_tickets)
```

```python
pending_tickets = df[
    df["Status"] == "Pending"
]["Ticket_ID"].nunique()

print("Pending Tickets:", pending_tickets)
```

```python
sla_breaches = (
    df["SLA_Status"] == "Breached"
).sum()

print("SLA Breaches:", sla_breaches)
```

```python
sla_met = (
    df["SLA_Status"] == "Met"
).sum()

sla_breached = (
    df["SLA_Status"] == "Breached"
).sum()

sla_compliance_rate = (
    sla_met / (sla_met + sla_breached)
) * 100

print(
    "SLA Compliance Rate:",
    round(sla_compliance_rate, 2),
    "%"
)
```

```python
average_resolution = df[
    df["Resolution_Hours"].notna()
]["Resolution_Hours"].mean()

print(
    "Average Resolution Time:",
    round(average_resolution, 2),
    "hours"
)
```

```python
average_satisfaction = (
    df["Customer_Satisfaction"].mean()
)

print(
    "Average Customer Satisfaction:",
    round(average_satisfaction, 2)
)
```

```python
category_volume = (
    df.groupby("Category")["Ticket_ID"]
    .nunique()
    .sort_values(ascending=False)
)

print(category_volume)
```

```python
category_volume.plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Ticket Volume by Category")
plt.xlabel("Category")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=45)
plt.show()
```

```python
category_breaches = (
    df[df["SLA_Status"] == "Breached"]
    .groupby("Category")["Ticket_ID"]
    .nunique()
    .sort_values(ascending=False)
)

print(category_breaches)
```

```python
category_sla = df.groupby("Category").agg(
    Total_Tickets=("Ticket_ID", "nunique"),
    Breached=("SLA_Status", lambda x: (x == "Breached").sum())
)

category_sla["Breach_Rate"] = (
    category_sla["Breached"] /
    category_sla["Total_Tickets"]
) * 100

category_sla.sort_values(
    "Breach_Rate",
    ascending=False
)
```

```python
priority_sla = df.groupby("Priority").agg(
    Total_Tickets=("Ticket_ID", "nunique"),
    Average_Resolution=("Resolution_Hours", "mean"),
    Breaches=("SLA_Status", lambda x: (x == "Breached").sum())
)

priority_sla["Breach_Rate"] = (
    priority_sla["Breaches"] /
    priority_sla["Total_Tickets"]
) * 100

priority_sla
```

```python
region_sla = df.groupby("Region").agg(
    Total_Tickets=("Ticket_ID", "nunique"),
    Average_Resolution=("Resolution_Hours", "mean"),
    Breaches=("SLA_Status", lambda x: (x == "Breached").sum())
)

region_sla["Breach_Rate"] = (
    region_sla["Breaches"] /
    region_sla["Total_Tickets"]
) * 100

region_sla.sort_values(
    "Breach_Rate",
    ascending=False
)
```

```python
channel_sla = df.groupby("Channel").agg(
    Total_Tickets=("Ticket_ID", "nunique"),
    Average_Resolution=("Resolution_Hours", "mean"),
    Breaches=("SLA_Status", lambda x: (x == "Breached").sum())
)

channel_sla["Breach_Rate"] = (
    channel_sla["Breaches"] /
    channel_sla["Total_Tickets"]
) * 100

channel_sla
```

```python
agent_analysis = df.groupby("Agent").agg(
    Total_Tickets=("Ticket_ID", "nunique"),
    Average_Resolution=("Resolution_Hours", "mean"),
    Breaches=("SLA_Status", lambda x: (x == "Breached").sum())
)

agent_analysis["Breach_Rate"] = (
    agent_analysis["Breaches"] /
    agent_analysis["Total_Tickets"]
) * 100

agent_analysis.sort_values(
    "Breach_Rate",
    ascending=False
)
```

```python
customer_analysis = df.groupby("Customer_Type").agg(
    Total_Tickets=("Ticket_ID", "nunique"),
    Average_Resolution=("Resolution_Hours", "mean"),
    Average_Satisfaction=("Customer_Satisfaction", "mean")
)

customer_analysis
```

```python
root_cause = df.groupby(
    ["Category", "Subcategory"]
).agg(
    Tickets=("Ticket_ID", "nunique"),
    Average_Resolution=("Resolution_Hours", "mean"),
    Breaches=("SLA_Status", lambda x: (x == "Breached").sum())
)

root_cause["Breach_Rate"] = (
    root_cause["Breaches"] /
    root_cause["Tickets"]
) * 100

root_cause.sort_values(
    "Breach_Rate",
    ascending=False
)
```

```python
df["Month"] = df["Created_Date"].dt.to_period("M").astype(str)
```

```python
monthly_volume = (
    df.groupby("Month")["Ticket_ID"]
    .nunique()
)

print(monthly_volume)
```

```python
monthly_volume.plot(
    kind="line",
    marker="o",
    figsize=(12, 5)
)

plt.title("Monthly Ticket Volume")
plt.xlabel("Month")
plt.ylabel("Tickets")
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_sla = df.groupby("Month").agg(
    Total=("Ticket_ID", "nunique"),
    Breaches=("SLA_Status", lambda x: (x == "Breached").sum())
)

monthly_sla["Breach_Rate"] = (
    monthly_sla["Breaches"] /
    monthly_sla["Total"]
) * 100

monthly_sla
```
